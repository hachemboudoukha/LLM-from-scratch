```python
from pathlib import Path
import urllib.request

repo_root = Path.cwd()
text_path = repo_root / 'the-verdict.txt'
if not text_path.exists():
    url = (
        'https://raw.githubusercontent.com/rasbt/'
        'LLMs-from-scratch/main/ch02/01_main-chapter-code/'
        'the-verdict.txt'
    )
    urllib.request.urlretrieve(url, text_path)

raw_text = text_path.read_text(encoding='utf-8')
print('Total characters:', len(raw_text))
print(raw_text[:120])
```

## 1. Tokenization

Tokenization means splitting text into smaller chunks such as words or punctuation. This is one of the first steps before feeding text to an LLM.


```python
import re

sample = 'Hello, world! This is a test.'
tokens = re.findall(r'\w+|[^\w\s]', sample)
print(tokens)
print('Total tokens:', len(tokens))
```

```python
import re
from collections import Counter

story_sample = raw_text[:1000]
story_tokens = re.findall(r'\w+|[^\w\s]', story_sample)
print(story_tokens[:40])
print('Sample token count:', len(story_tokens))

words = re.findall(r'\b\w+\b', raw_text.lower())
counter = Counter(words)
print('Top 10 words:')
for word, count in counter.most_common(10):
    print(word, count)
```

## 2. Embeddings intuition

Embeddings turn tokens into vectors. Similar words usually end up close together in this vector space.

This is the main idea behind language models: they learn useful numerical representations of text.


```python
import numpy as np

vectors = {
    'king': np.array([1.0, 0.9, 0.2]),
    'queen': np.array([0.9, 1.0, 0.4]),
    'apple': np.array([0.1, 0.2, 1.0]),
    'orange': np.array([0.2, 0.3, 0.9]),
}

for a, v1 in vectors.items():
    for b, v2 in vectors.items():
        if a < b:
            dist = np.linalg.norm(v1 - v2)
            print(f'{a} vs {b}: {dist:.3f}')
```
